# Project Problem
Although standardising geological and geotechnical logging across mines remains important, it requires extensive multi-operation datasets. This project therefore uses data from a single operation to assess whether geotechnical parameters can predict stope stability. The findings may support future efforts to develop more objective and standardised geotechnical characterisation methods.

Underground stope stability is a critical aspect of mine design and safety. Traditional stability assessment methods, such as the Mathews Stability Graph, use empirical relationships between rock mass quality and stope geometry to estimate stability. While these methods have proven valuable, they may not fully capture complex interactions between geotechnical and geometric parameters observed in historical stope performance data.

# Intended AI/ML Outcome

The aim of this project is to develop a machine learning classification model capable of predicting stope stability from geotechnical and geometric parameters. Historical stope case data will be used to train models that classify stopes as either Stable or Unstable based on observed ELOS performance.

This project also serves as an initial step toward improving the objectivity and consistency of geotechnical assessment through data-driven methods.

# Data Required

The project requires historical stope case histories containing both geotechnical and geometric information.

The required variables include:

## Geotechnical Variables

- Modified Stability Number (N')
- Modified Q value (Q')
- A factor
- B factor
- C factor

## Geometric Variables

- Hydraulic Radius (HR)
- Top Stope Length
- Bottom Stope Length
- North Wall Height
- South Wall Height

## Stability Measure

- Equivalent Linear Overbreak/Slough (ELOS)

ELOS is used to determine the observed stability class:

- Stable: ELOS < 0.6 m
- Unstable: ELOS ≥ 0.6 m

# Dataset

The dataset consists of historical underground stope case studies collected as part of ongoing MEng research into stope stability prediction.

Each record represents an individual stope surface and contains geotechnical parameters, geometric measurements, and observed ELOS values. The dataset is used to investigate whether machine learning algorithms can identify patterns associated with stable and unstable stopes.
