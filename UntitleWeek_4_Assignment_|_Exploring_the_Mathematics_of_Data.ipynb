#Descriptive Statistics and Correlation Analysis for Stope Stability Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Excel file
df = pd.read_excel(
    "/content/STOPE ID.xlsx",
    sheet_name="Stope Data"
)

df.head()
```

```python
variables = [
    "Q'",
    "N'",
    "HR",
    "ELOS (m)"
]

df[variables].head()
```

```python
for col in variables:

    mean = df[col].mean()
    median = df[col].median()
    value_range = df[col].max() - df[col].min()
    variance = df[col].var()
    std = df[col].std()

    print(f"\n{'='*40}")
    print(f"Variable: {col}")
    print(f"Mean: {mean:.3f}")
    print(f"Median: {median:.3f}")
    print(f"Range: {value_range:.3f}")
    print(f"Variance: {variance:.3f}")
    print(f"Standard Deviation: {std:.3f}")
```

```python
covariance = df["N'"].cov(df["ELOS (m)"])
correlation = df["N'"].corr(df["ELOS (m)"])

print("Covariance =", covariance)
print("Correlation =", correlation)
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x="N'",
    y="ELOS (m)"
)

plt.title("Relationship Between N' and ELOS")
plt.xlabel("Modified Stability Number (N')")
plt.ylabel("ELOS (m)")
plt.show()
```

## What do the statistics tell us about the data?


The average Q' value is 7.41, with a standard deviation of 2.99, indicating moderate variation in rock mass quality across the stopes. The average N' value is 16.26 with a standard deviation of 8.33, suggesting a wide range of stability conditions represented in the dataset.

Hydraulic Radius (HR) has a mean of 7.51 m and a relatively small standard deviation of 0.69 m, indicating that stope sizes are comparatively consistent. In contrast, ELOS exhibits substantial variability, with a mean of 0.64 m, a median of 0.34 m, and a standard deviation of 0.77 m. The large difference between the mean and median suggests that the ELOS distribution is positively skewed, with a small number of high-overbreak stopes increasing the average value

## What does the correlation tell us about the relationship between the two variables?

The covariance between N' and ELOS is -1.598, while the correlation coefficient is -0.249.

The negative correlation indicates that higher N' values tend to be associated with lower ELOS values. This relationship is consistent with geotechnical theory because larger N' values generally represent more favourable stability conditions.

However, the correlation magnitude of -0.249 is relatively weak. This suggests that N' alone does not strongly explain variations in ELOS. Other factors, such as hydraulic radius, stope geometry, geological structures, stress conditions, and local rock mass characteristics, also influence stope performance.

The scatter plot supports this conclusion, as the data points are widely dispersed rather than forming a clear linear trend.

## Does correlation imply causation?

No. Correlation does not imply causation.

Although N' and ELOS show a weak negative relationship, this does not prove that changes in N' directly cause changes in ELOS. Stope stability is influenced by many interacting geological, geotechnical, and operational factors.

Correlation simply indicates that two variables tend to vary together. Determining causation requires additional evidence, physical understanding of the system, and more detailed analysis.
