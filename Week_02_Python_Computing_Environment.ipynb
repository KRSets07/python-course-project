## Temperature Conversion Program

```python
# Week 2 Assignment: Temperature Conversion

# Function to convert Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit


# Ask the user to enter a temperature
temperature = float(input("Enter a temperature in °C: "))

# Convert the temperature to Fahrenheit
converted_temperature = celsius_to_fahrenheit(temperature)

# Display the converted temperature
print(f"{temperature}°C is equal to {converted_temperature:.2f}°F")


# Specify the temperature threshold
threshold = 30

# Check whether the temperature is above or below the threshold
if temperature > threshold:
    print(f"The temperature is above the threshold of {threshold}°C.")
elif temperature < threshold:
    print(f"The temperature is below the threshold of {threshold}°C.")
else:
    print(f"The temperature is equal to the threshold of {threshold}°C.")


# Convert several temperatures using a loop
temperatures = [10, 20, 25, 30, 35, 40]

print("\nTemperature conversions:")

for temp in temperatures:
    fahrenheit = celsius_to_fahrenheit(temp)
    print(f"{temp}°C = {fahrenheit:.2f}°F")
```
